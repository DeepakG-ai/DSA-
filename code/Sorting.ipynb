```python
#Selection Sort
arr = [13,46,24,52,20,9]
def selection_sort(arr,n):
    for i in range(n-1):
        min =i
        for j in range(i+1,n):
            if arr[j]< arr[min]:
                min = j
        arr[min],arr[i] = arr[i],arr[min]
    return arr
print(selection_sort(arr,len(arr)))
 
```

```python
arr = [13,46,24,52,20,9]
def bubble_sort(arr,n):
    for i in range(n-1):
        swapped = False
        for j in range(n-1-i):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
                swapped = True
        if not swapped:
            break
    return arr

print(bubble_sort(arr,len(arr)))
```

```python
arr = [13,46,24,52,20,9]
def insertion_sort(arr,n):
    for i in range(1,n):
        key = arr[i]
        j=i-1
        while (j>=0) and key<arr[j]:
            arr[j+1]=arr[j]
            j-=1
        arr[j+1]=key
    return arr
print(insertion_sort(arr,len(arr)))
```

```python
def merge(arr, low, mid, high):
    # Create two temporary arrays
    left = arr[low:mid+1]
    right = arr[mid+1:high+1]

    # Indices for left, right, and merged arrays
    i = j = 0
    k = low

    # Merge the two arrays
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            arr[k] = left[i]
            i += 1
        else:
            arr[k] = right[j]
            j += 1
        k += 1

    # Add remaining elements from left array
    while i < len(left):
        arr[k] = left[i]
        i += 1
        k += 1

    # Add remaining elements from right array
    while j < len(right):
        arr[k] = right[j]
        j += 1
        k += 1


def merge_sort(arr, low, high):
    if low < high:
        # Find the middle of the array
        mid = (low + high) // 2

        # Recursively sort left and right halves
        merge_sort(arr, low, mid)
        merge_sort(arr, mid + 1, high)

        # Merge the sorted halves
        merge(arr, low, mid, high)


# Example usage
arr = [38, 27, 43, 3, 9, 82, 10]
print("Original array:", arr)
merge_sort(arr, 0, len(arr) - 1)  # Sort the array
print("Sorted array:", arr)
```

```python
#Quick Sort
arr = [38, 27, 43, 3, 9, 82, 10]
def quick_sort(arr):
    if len(arr)<=1:  #Base Case
        return arr
    
    p=arr[-1]   #Pivot
    L = [x for x  in arr[:-1] if x<=p]
    R = [x for x  in arr[:-1] if x>p]

    L = quick_sort(L)
    R = quick_sort(R)

    return L+[p]+R
quick_sort(arr)

    
```

```python
#Quick Sort
def partition(arr,low,high):
    pivot = arr[high]
    i = low - 1  # pointer for smaller elements  initially i= -1
    
    for j in range(low,high):
        if arr[j]<= pivot:
            i+=1
            arr[i],arr[j] = arr[j],arr[i]
    arr[i+1],arr[high] = arr[high],arr[i+1]
    return i+1

def quick_sort(arr,low,high):
    if low<high:
        pi = partition(arr,low,high)

        quick_sort(arr,low,pi-1)
        quick_sort(arr,pi+1,high)

arr = [13,46,24,52,20,9]
low = 0
high = len(arr)-1
quick_sort(arr,low,high)
print("Sorted Array",arr)
```
