```python
#Linear Search 
a = [-3,-1,0,1,4,7]
s=4
for i in range(0,len(a)):
    if a[i]==s:
        print(i)

# if 4 in A: print(True)  this is also work
```

```python
#Binary Search
#Time O(log n)
#Space O(1)
def binary_search(a,s):
    a=[-1,4,5,7,8,14,56,78]
    s=56
    l,r=0,len(a)-1
    while l<=r:
        mid = (l+r)//2
        if a[mid]==s:
            print(mid)
            break
        elif l<a[mid]:
            l=mid+1
        else:
            r=mid-1
    else:
        print(-1)

binary_search(a,s)
```

```python
#Binary Search
class TreeNode:
    def __init__(self,val,left=None,right=None):
        self.val=val
        self.left = left
        self.right = right

    def __str__(self):
        return str(self.val)
    
    
```

```python
#    1
#  2    3
#4  5   10
```

```python
A = TreeNode(1)
B = TreeNode(2)
C = TreeNode(3)
D = TreeNode(4)
E = TreeNode(5)
F = TreeNode(10)

A.left = B
A.right = C
B.left = D
B.right = E
C.left = F
print(A)
```

```python
#Recursive Pre-Order Traversal (DFS) Time : O(n), Space : O(n)
def pre_order(node):
    if not node:
        return 
    
    print(node)
    pre_order(node.left)
    pre_order(node.right)

pre_order(A)
```

```python
#Recursive In-Order Traversal (DFS) Time : O(n), Space : O(n)
def in_order(node):
    if not node:
        return 
    
    in_order(node.left)
    print(node)
    in_order(node.right)

in_order(A)
```

```python
#Recursive Pre-Order Traversal (DFS) Time : O(n), Space : O(n)
def post_order(node):
    if not node:
        return 
    
    post_order(node.left)
    post_order(node.right)
    print(node)

post_order(A)
```

```python
#Iterative Pre Order 
def pre_order_iterative(node):
    stk=[node]

    while stk:
        node = stk.pop()
        print(node)
        if node.right:  stk.append(node.right)
        if node.left:   stk.append(node.left)

pre_order_iterative(A)
```

```python
# BFS - Level Order Traversal
from collections import deque

def level_order(node):
    q= deque()
    q.append(node)

    while q:
        node = q.popleft()
        print(node)
        if node.left: q.append(node.left)
        if node.right: q.append(node.right)

level_order(A)
```

```python
#check if value exists (DFS)
def search(node,target):
    if not node:
        return False
    
    if node.val == target:
        return True
    return search(node.left,target) or search(node.right,target)

search(A,2)
```
