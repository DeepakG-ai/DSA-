```python
def isArmStrong(number):
    k=len(str(number)) #length of number 
    sum=0 
    n = number
    while n>0:
        ld=n%10 #extract last digit
        sum+=ld**k #power of digit
        n=n//10 #remove the last digit
    return sum==number #check if sum = original number
def main():
    number=123
    if isArmStrong(number):
        print(number,"is ArmStrong Number")
    else:
        print(number,"is not ArmStrong Number")
if __name__=="__main__":
    main()
```

```python
#check palindrome 
def palindrome(n):
    revNum =0
    dup = n
    while n>0:
        ld = n%10
        revNum = (revNum*10)+ld
        n=n//10
    if dup==revNum:
        return True
    else:
        return False
def main():
    number=121
    if palindrome(number):
        print(number,"is Palindrome Number")
    else:
        print(number,"is not Palindrome Number")
if __name__=="__main__":
    main()



```

```python
#parametrised sum of first n numbers 
def f(i,sum):
    if i<1:
        return sum
    else:
         return f(i-1,sum+i)
if __name__=="__main__":
    n=5
    print(f(n,0))
```

```python
#function print sum of first n numbers
def f(n):
    if (n==0): #Base Condition
        return 0
    else:
        return n+f(n-1) #recurisive   5+f(4), 9+f(3), 12+f(2),14+f(1),15+f(0) then base condition true 
if __name__=="__main__":
    n=5
    print(f(n)) #printing the sum
```

```python
#factorial of N
def f(n):
    if (n==1): #Base condition
        return 1                #TC-> O(n)
    else:                       #SC->O(n)
        return n*f(n-1)  
    
if __name__=="__main__":
    n=5
    print(f(n)) #printing the sum
    
```

```python
def f(arr,l,r):
    if l>=r:
        return
    
    arr[l],arr[r]=arr[r],arr[l]
    f(arr,l+1,r-1)

arr=[2,3,5,7,4,3,1]
f(arr,0,len(arr)-1)
print(arr)
```
