```python
#Heap 
A = [-4,3,1,0,2,5,10,8,12,9]

import heapq
heapq.heapify(A)

print(A)
```

```python
#Push Operation
heapq.heappush(A, 6)

A
```

```python
#Heap Pop 
minn = heapq.heappop(A)

A,minn
```

```python
#Heap Sort
#Time -> O(n logn) Space ->  O(n)
def heapsort(arr):
    heapq.heapify(arr)
    n=len(arr)
    new_list = [0]*n

    for i in range(n):
        minn= heapq.heappop(arr)
        new_list[i]=minn
    
    return new_list
heapsort([1,3,5,7,9,2,4,6,8,0])
```

```python
#Max Heap
B=[-4,3,1,0,2,5,10,8,12,9]
n=len(B)

for i in range(n):
    B[i]=-B[i]
heapq.heapify(B)

B
```

```python
#Largest Element 
largest = -heapq.heappop(B)
largest
```

```python
#Build Heap from Scratch
C = [-4,3,1,0,2,5,10,8,12,9]
heap=[]
for x in C:
    heapq.heappush(heap,x)
    print(heap)
```

```python
#Putting tupples of item on Heap
D=[5,4,3,5,4,3,5,5,4]

from collections import Counter
counter = Counter(D)
counter
```

```python
heap=[]

for k,v in counter.items():
    heapq.heappush(heap,(v,k))

heap
```
