```python
class Solution:
    def twoSum(self,nums,target):
        d={}
        for i in range(0, len(nums)):
            d[nums[i]]=i        #{2:0,7:1,11:2,15:3}
        for i in range(0,len(nums)):
            x=target - nums[i]               #x+nums[0]=9  2+7 which are in 0 and 1 index
            if x in d and i !=d[x]:
                return [i,d[x]]
        return None
s= Solution()
result = s.twoSum([2, 7, 11, 15], 9)  # Example input
print(result)  # Expected output: [0, 1]
```

2) Longest Substring without repeating characters


```python
s=["abcabcdbb"]
def lengthOfLongestSubstring(s: str) -> int:
    char_set = set()
    left = 0
    res=0
    
    for right in range(len(s)):
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1
        char_set.add(s[right])
        res=max(res, right - left + 1)
    
    return res

s = "abcabcbb"  # example string
print("Length of longest substring without repeating characters:", lengthOfLongestSubstring(s))

```

3) Product of array except self 

```python
def productExceptSelf(nums):
    n = len(nums)
    
    # Initialize the answer array
    answer = [1] * n
    
    
    # Calculate products of elements to the left of each index
    left_product = 1    
    for i in range(n):
        answer[i] = left_product #answer[i]=[1,1,1,1] (initialized as all 1s)       num[i]=[1,2,3,4]
        left_product *= nums[i] #left_product *= nums[0] → left_product = 1  * 1 (num[i]) = 1
                             #i=2-->left_product = 2 * 3 = 6, i=3-->answer = [1, 1, 2, 6] left_product = 6 * 4 = 24
  
    # Calculate products of elements to the right of each index
    right_product = 1
    for i in range(n - 1, -1, -1): # iterate through the array in reverse order, from the last index to the first.
        answer[i] *= right_product 
                                        #n-1 last index. -1 end point. -1 step part.
        right_product *= nums[i]
         
    
    return answer

# Example usage:
nums = [1, 2, 3, 4]
print(productExceptSelf(nums))
             # Output: [24, 12, 8, 6]
```

4.Maximum Subarray


```python
class solution:
    def maxSubArray(self,nums): #kadanes algorithm
        maxsub=nums[0]
        cursum=0

        for i in nums:
            if cursum<0:
                cursum=0
            cursum+=i
            maxsub=max(maxsub,cursum)
        return maxsub
    
s=solution()
s.maxSubArray([-2,1,-3,4,-1,2,1,-5,4])
```

5. Merged intervals


```python
def merge(intervals):
    # If there are no intervals, return an empty list
    if len(intervals) == 0:
        return []

    # Step 1: Sort the intervals based on the starting time
    intervals.sort()

    # Step 2: Initialize the merged list with the first interval
    merged_intervals = [intervals[0]]

    # Step 3: Go through the rest of the intervals
    for i in range(1, len(intervals)):
        # Get the last interval in the merged list
        last_merged = merged_intervals[-1]
        
        # Get the current interval we're looking at
        current_interval = intervals[i]
        
        # Step 4: Check if the current interval overlaps with the last merged one
        if current_interval[0] <= last_merged[1]:
            # If they overlap, merge them by updating the end of the last merged interval
            last_merged[1] = max(last_merged[1], current_interval[1])
        else:
            # If they don't overlap, simply add the current interval to the merged list
            merged_intervals.append(current_interval)

    # Step 5: Return the merged list of intervals
    return merged_intervals
intervals = [[1,3],[2,6]]

result = merge(intervals)
print(result)
```

```python
#iterative method
class ListNode:
    def __init__(self,value=0,next=None):
        self.value=value
        self.next=next
    def reverseList(head:'ListNode') -> 'ListNode':
        cur=head
        prev=None

        while cur:          #head=[1 2 3 4 5]     o/p=[5 4 3 2 1]  
            temp=cur.next                               #cur=2 temp=3 cur.next=1 prev=2 cur=3 exhange cur and prev using temporarly variable
            cur.next=prev
            prev=cur
            cur=temp
        
        return prev

```

3. Detect a cycle in a Linked List

```python
class solutino:
    def hascycle(self,head:ListNode)-> bool:
        dummy= ListNode()
        dummy.next=head
        slow=fast=dummy

        while fast and fast.next:
            fast=fast.next.next
            slow=slow.next
         
            if slow is fast:
                return True
            
        
        return False
```

4. Remove the nth node from the end of list


```python
class solution:
    def removeNthFromEnd(self, head:'ListNode',n:int)-> 'ListNode':
        dummy=ListNode(0,head)
        left=dummy
        right=head

        while n>0 and right:
            right=right.next
            n-=1

        while right:
            left=left.next
            right=right.next

        left.next=left.next.next
        return dummy.next
        
```

1. Valid Parenthesis

```python
class solution:
    def isValid(self,s:str):
        stack=[] #empty list
        closeToOpen={")":"(","}":"{","]":"["}

        for i in s:
            if i in closeToOpen:
                if stack and stack[-1]==closeToOpen[i]:
                    stack.pop()
                else:
                    return False
            else:
                stack.append(i)

        return not stack 
    
c=solution()
result=c.isValid("([{[]}])")
print(result)
```

4. Largest Rectangle of Histogram

```python
class solution:
    def LargestRecArea(self,height:[int])-> int:
        n=len(height)
        stk=[]
        max_area=0

        for i , height in enumerate(height):
            start =i
            while stk and height < stk[-1][0]:
                h,j=stk.pop()
                w=i-j
                a=h*w
                max_area=max(max_area,a)
                start=j
            stk.append((height,start))

        while stk:
            h,j=stk.pop()
            w=n-j
            max_area=max(max_area,h*w)

        return max_area
```

```python
def climb(n):
    if n==1:
        return 1
    elif n == 2:
        return 2
    dp=[0]*(n+1)
    dp[1]=1
    dp[2]=2
    for i in range(3,n+1):
        dp[i]=dp[i-1]+dp[i-2]

    return dp[n]

print(climb(5))
```

2. Longest increasing subsequences

```python
class solution:
    def lengthofLIS (self,nums):
        n= len(nums)
        dp = [1]*n

        for i in range(1,n):
            for j in range(i):
                if nums[i]>nums[j]:
                    dp[i]=max(dp[i],dp[j]+1)
        return max(dp)

s = solution()
s.lengthofLIS([4,5,2,6,7])
```

3. longest comman subsequences

```python
class solution:
    def Longestcommansubsequence(self,text1,text2):
        #Top Down DP (Memorization)
        m,n = len(text1),len(text2)
        
        def longest(i,j):
            if i == m or j ==n:
                return 0
            elif text1[i]==text2[j]:
                return 1+longest(i+1,j+1)
            else:
                return max(longest(i,j+1),longest(i+1,j))
        return longest(0,0)
    
s = solution()
s.Longestcommansubsequence("rajushekar","raj")
```

```python
class solution:
    def Longestcommansubsequence(self,text1,text2):
        #Bottom up DP (Memorization)
        m,n = len(text1),len(text2)
        dp=[[0] *(n+1) for _ in range(n+1)]

        for i in range(1,n+1):
            for j in range(1,n+1):
                if text1[i-1] == text2[j-1]:
                    dp[i][j]=dp[i-1][j-1] +1 
                else:
                    dp[i][j] = max(dp[i-1][j],dp[i][j-1])
        
        return dp[m][n]
```

4. Coin Change

```python
class solution:
    def coinchange(self,coins,amount:int)->int:
        dp = [amount+1]*(amount +1)
        dp[0] = 0

        for a in range(1,amount+1):
            for c in coins:
                if a-c >=0:
                    dp[a] = min(dp[a],1+dp[a-c])
        
        return dp[amount] if dp[amount] != amount+1 else -1
    
s = solution()
s.coinchange([1,4,3,5],7)
```

2. Quick Sort 

```python
class solution:
    def quicksort(self,nums):
        l,r= 0,len(nums)-1
        i=0

        def swap(i,j):
            temp = nums[i]
            nums[i]=nums[j]
            nums[j]=temp

        while i<=r:
            if nums[i]==0:
                swap(l,i)
                l+=1
            elif nums[i]==2:
                swap(i,r)
                r-=1
                i-=1
            i+=1 
```

3.kth largest element in array

```python
class solution:
    def findKthLargest(self,nums,k):
        k=len(nums)-k

        def quickSelect(l,r):
            pivot,p=nums[r],l
            for i in range(l,r):
                if nums[i]<pivot:
                    nums[p],nums[i] = nums[i],nums[p]
                    p +=1
            nums[p],nums[r]=nums[r],nums[p]

            if p>k: return quickSelect(l,p-1)
            elif p<k: return quickSelect(p+1,r)
            else:
                return nums[p]
        
        return quickSelect(0,len(nums)-1)
            
    s=solution()
    nums,k=[1,2,4,3,5,6],2
    print(s.findKthLargest(nums,k))

```

4. find peak element 

```python
def findPeakElement(nums):
    l,r= 0,len(nums)-1    #l,r=0,6
    
    while l<r:
        mid = (l+r)//2
       

        if nums[mid]<nums[mid+1]:
            l=mid+1
        else:
            r=mid
    return l

nums=[1,2,1,3,5,6,4]
print(findPeakElement(nums))

```
