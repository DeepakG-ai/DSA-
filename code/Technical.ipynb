```python
#1. Find the Subarray with a Given Sum
def subarray_with_sum(arr,target):
    result = []
    for i in range(len(arr)):
        current_sum = 0
        subarray = []
        for j in range(i,len(arr)):
            current_sum+=arr[j]
            subarray.append(arr[j])
            if current_sum==target:
                result.append(subarray[:])
    return result

arr=[3,4,-7,1,3,3,1,-4]
target = 0
output = subarray_with_sum(arr,target)  
print("subarray with sum",target,":",output)
```

```python
def rotate_array(arr, K):
    N = len(arr)
    K = K % N  # Reduce K to avoid unnecessary full rotations
    return arr[-K:] + arr[:-K]

# Example usage:
arr1 = [10, 20, 30, 40, 50]
K1 = 4
result1 = rotate_array(arr1, K1)
print(result1)  # Output: [40, 50, 10, 20, 30]

arr2 = [10, 20, 30, 40]
K2 = 1
result2 = rotate_array(arr2, K2)
print(result2)  # Output: [40, 10, 20, 30]

        
```

```python
def reformat_camel_case(s):
    result = []
    for char in s:
        if char.isupper():
            result.append(' ')
        result.append(char.lower())
    return ''.join(result).strip()

# Example usage:
print(reformat_camel_case("ThisIsAnAutomationEra"))  # Output: "this is an automation era"
print(reformat_camel_case("HeyYou"))                 # Output: "hey you"
```

```python
def reverse_vowels(s):
    vowels = set('aeiouAEIOU')
    s = list(s)
    left, right = 0, len(s) - 1

    while left < right:
        if s[left] not in vowels:
            left += 1
        elif s[right] not in vowels:
            right -= 1
        else:
            s[left], s[right] = s[right], s[left]
            left += 1
            right -= 1

    return ''.join(s)

# Example usage:
print(reverse_vowels("hello"))  # Output: "holle"
print(reverse_vowels("game"))   # Output: "gmea"
```
