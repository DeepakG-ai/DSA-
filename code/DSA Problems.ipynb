1. Two Sum 

```python
def TwoSum(arr,target):
    for i in range(0,len(arr)-1):
        for j in range(i+1,len(arr)-1):
            sum=arr[i]+arr[j]
            if target==sum:
                return i,j
    
    return -1
            
TwoSum([2,8,5,7,9],7)
#Time Complexity O(n)
```

```python
def TwoSum(nums, target):
    seen = {}  # value: index
    for i, num in enumerate(nums):
        diff = target - num
        if diff in seen:
            return [seen[diff], i]
        seen[num] = i
TwoSum([2,8,5,7,9],7)
```

2. Reverse an array 

```python
arr = [1,3,5,2,6,8]

def reverse(l,r):
    if l>=r:
        return arr
    else:
        arr[l],arr[r]=arr[r],arr[l]
        reverse(l+1,r-1)

reverse(0,len(arr)-1)
print(arr)
```

3. Finding Missing Number 1 - N

```python
def Sum(nums):
    n=len(nums)
    total_Sum = n*(n+1)//2
    sum1=0
    for i in range(n-1):
        sum1+=nums[i]
    return total_Sum-sum1
Sum([1,2,4,5])

```

4. Find the Largest and Smallest element in array

```python
def Largest(arr):
    largest=arr[0]
    for i in range(len(arr)):
        if arr[i] > largest:
            largest = arr[i]
    return largest

def Smallest(arr):
    smallest =arr[0]
    for i in range(len(arr)):
        if arr[i]<smallest:
            smallest = arr[i]
    return smallest

print(Largest([2,4,56,7,34,95]))
print(Smallest([2,4,56,7,34,95]))
```

6. Moves Zeros at the end of the array 

```python
def move_Zeros(arr):
    n = len(arr)
    left = 0
    for right in range(n):
        if arr[right]!=0:
            arr[left],arr[right]=arr[right],arr[left]
            left+=1
    return arr

move_Zeros([1,0,3,2,0,0,5])
```

7. Check if an array is Sorted 

```python
arr = [2,4,3,1,5,4]
def isSorted(arr):
    for i in range(1, len(arr)):
        if arr[i] < arr[i - 1]:
            return False
    return True

isSorted(arr)
```

8. Union and Intersection of two array

```python
arr1 = [1,2,3,1,2,5]
arr2 = [1,2,5,4]

def Union(arr1,arr2):
    result = set()

    for i in arr1:
        result.add(i)
    for i in arr2:
        result.add(i)

    return sorted(result)
Union(arr1,arr2)
```

```python
result = set(arr1).intersection(set(arr2))
print(list(result))
```

9. Rotate an array 

```python
#Right Rotate 
def rotate(nums,k):
    n = len(nums)
    k= k%n

    def reverse(l,r):
        while l<r:
            nums[l],nums[r]=nums[r],nums[l]
            l+=1
            r-=1

    reverse(0,n-1)
    reverse(0,k-1)
    reverse(k,n-1)

    return nums

result = rotate([1,2,3,4,5,6,7],3) 
print(result)
```

```python
#left  Rotate 
def rotate(nums,k):
    n = len(nums)
    k=k%n

    def reverse(l,r):
        while l<r:
            nums[l],nums[r]= nums[r], nums[l]
            l+=1
            r-=1

    reverse(0,k-1)
    reverse(k,n-1)
    reverse(0,n-1)

    return nums


result1 = rotate([1,2,3,4,5,6,7],3) 
print(result1)
```

```python
# Rotate an array 
def leftRotate(arr,k):
    n=len(arr)
    k%=n
    return arr[k:]+arr[:k]

def rightRotate(arr,k):
    n=len(arr)
    k=k%n
    return arr[-k:]+arr[:-k]

print(leftRotate([1,2,3,4,5],3))
print(rightRotate([1,2,3,4,5],3))
```

11. Best time to buy and sell stock. 


```python
def MaxProfit(prices):
    min_price = prices[0]
    max_profit = 0

    for price in prices:
        if price<min_price:
            min_price=price
        else:
            max_profit=max(max_profit,price-min_price)
            
    return max_profit
MaxProfit([7,1,5,3,4,6])
```

12. Maximun subarray sum

```python
def maxSubArray(nums): #kadanes algorithm
        maxsub=nums[0]
        cursum=0

        for i in nums:
            if cursum<0:
                cursum=0
            cursum+=i
            maxsub=max(maxsub,cursum)
        return maxsub
maxSubArray([-2,1,-3,4,-1,2,1,-5,4])
```
