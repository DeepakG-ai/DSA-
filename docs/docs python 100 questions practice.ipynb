````python
from IPython.lib.guisupport import is_event_loop_running_qt4

a = [1,2,3]
b = a
a.append(4)  #list are mutable. append is operation. so python can easily modify without creating the new object.
print(b)
b.append(9)
print(b)
print(a)  # a and b are pointing to same memory. b is doesn't created the another list copy of a
print(id(a)) # both are same memory
print(id(b))
print ( a is b)
````

````python
a = [1, 2, 3]
b = a.copy()   #here, b is copy of a. not pointing to same memory
a.append(4)
print(b)
print(a)

print(id(a))
print(id(b))
print ( a is b)
````

````python
x = 5
y = x # at this point, x and y both are pointing same memory
print(x is y)
x = x+1 #so here, x is integer which is immutable, so python can't change or modify the value 6. so it creats new object. so y and x are different at this stage
print(y)
print(x)  # x value is increased.
print (x is y)
````

| Type                         | Can it be edited in place? | What happens on "change"                       |
| ---------------------------- | -------------------------- | -----------------
| int, float, str, tuple, bool | No (immutable)             | New object created, variable re-pointed to it  |
| list, dict, set              | Yes (mutable)              | Same object edited directly, address unchanged |

````python
"""Yes — here are clear examples.

## 1) Return a new value
Use `return` when the function creates a new object/value and you want the caller to use it.

```python
def add_one(x):
    return x + 1

a = 5
b = add_one(a)

print(a)  # 5
print(b)  # 6
```

Here `a` does not change. The function gives back a new result, so `return` is needed.

Another example:

```python
def upper_text(s):
    return s.upper()

name = "max"
new_name = upper_text(name)

print(name)      # max
print(new_name)  # MAX
```

## 2) Mutate in place
No `return` is needed when the function changes the same mutable object directly.

```python
def add_item(lst):
    lst.append(4)

a = [1, 2, 3]
add_item(a)

print(a)  # [1, 2, 3, 4]
```

The list itself changed, so the function did not need to return anything.

Another example:

```python
def update_count(d):
    d["count"] += 1

data = {"count": 0}
update_count(data)

print(data)  # {'count': 1}
```

## 3) Same idea, different behavior
Compare these two:

```python
def new_list(lst):
    return lst +  [courses.cs.washington](https://courses.cs.washington.edu/courses/cse331/11sp/lectures/slides/07-mutation.pdf)

def mutate_list(lst):
    lst.append(4)
```

Usage:

```python
a = [1, 2, 3]

b = new_list(a)
print(a)  # [1, 2, 3]
print(b)  # [1, 2, 3, 4]

mutate_list(a)
print(a)  # [1, 2, 3, 4]
```

## 4) Important interview line
You can say:

- Return a value when you create a **new object**.
- Do not return a value when you **mutate an existing mutable object in place**.

## 5) Tiny warning
Some methods both mutate and return something else, like `pop()`.

```python
a = [1, 2, 3]
x = a.pop()
print(a)  # [1, 2]
print(x)   # 3
```

So the rule is not “never return with mutation,” but rather:
- if the function’s purpose is mutation, usually return `None`
- if the function’s purpose is transformation, return the new object

Would you like 10 interview-style examples on this exact topic?"""
````

````python
from collections import Counter

users = ["ram", "sam", "ram", "tom", "sam", "ram"]

# Count frequency of each item
counts = Counter(users)

# 1. Unique names
unique_names = list(counts.keys())  # or set(users)

# 2. Count of each name
name_counts = dict(counts)

# 3. Most common name
most_common_name, frequency = counts.most_common(1)[0]

print("Unique names:", unique_names)
print("Counts:", name_counts)
print(f"Most common name: {most_common_name} (appears {frequency} times)")
````

````python
result = {}
for word in ["apple", "banana", "kiwi"]:
    result[word] = len(word)

print(result)
````

````python
words = ["apple", "banana", "kiwi"]

result = {word: len(word) for word in words}

print(result)
"""Syntax Breakdown
Plaintext
{ key_expr : value_expr  for item in iterable }
     │           │              │
   word      len(word)    for word in words
key_expr (word): What you want to use as the dictionary key.

value_expr (len(word)): What you want to store as the corresponding value.

for item in iterable (for word in words): The loop that iterates over your input list."""
````
