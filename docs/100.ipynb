````python
#accessing and changing the object data
class calcy:
    def __init__(self,a,b):
        self.a = a
        self.b = b

    def division(self):
        return self.a/self.b

    def multiplication(self,new_a):
        self.a = new_a
        return self.a*self.b

cly = calcy(5,10)
print(cly.division())  #5/10 = 0.5

print(cly.multiplication(12))  #12*10 = 120 not 5*10. because it is changing the object data.

````

````python
class Car:
    def __init__(self,name,color,speed):
        self.name = name
        self.color = color
        self.speed = speed

    def acceleration(self,acc_speed): #changed the object data of speed here
        self.speed = self.speed+ acc_speed
        return self.speed

    def get_car_info(self):
        return f"\n Car name :{self.name} \n Car color :{self.color} \n Car speed:{self.speed}"

car1 = Car("BMW","Black",220)
car2 = Car("Audi","Red",190)
car3 = Car("Fortuner","White",180)
car4 = Car("Supra","Green",300)

print(car1.acceleration(56))
print(car1.get_car_info())
print(car2.get_car_info())
print(car3.get_car_info())
print(car4.get_car_info()) #before acceleration
print(car4.acceleration(80))
print(car4.get_car_info())  #after the acceleration


````

````python
# Static Method

"""
A method inside the that does not access or read object data [self]
Inside class it behaves like regular function
"""

class static:
    def __init__(self,a,b):
        self.a = a
        self.b = b

    def add(self):
        return self.a + self.b

    @staticmethod
    def addition(x,y):
        return x+y


s = static(4,5)
print(s.add())
print(s.addition(7,8))
````

````python
def add(item, bucket=None):
    if bucket is None:
        bucket = []  # Creates a fresh list object on every function call
    bucket.append(item)
    return bucket

print(add("a"))
print(add("b"))
"""None is immutable, making it safe to use as a default argument.

On the first call (add("a")), bucket is None, so bucket = [] allocates a new list instance in memory.

On the second call (add("b")), bucket is None again, so Python creates another separate list in memory instead of modifying the previous one.
"""
````

````python
## Positional argument

"""A **positional argument** is matched to a parameter based on its **position/order**, not its name.

```python
def greet(name, age):
    print(name, age)

greet("Max", 25)
```

Here, `"Max"` goes to `name` and `25` goes to `age` simply because of their **order**. If you swap the order, the meaning changes:

```python
greet(25, "Max")  # name=25, age="Max" — wrong values, same positions
```

## Keyword argument

A **keyword argument** is passed as a `name=value` pair, so it's matched by **name**, not position.

```python
greet(age=25, name="Max")
```

This works correctly even though the order is swapped, because you explicitly told Python which value belongs to which parameter.

## Your instinct: "keyword and key-value pair" — correct

Yes, that's exactly right. A keyword argument **is** a key-value pair:
- `key` = the parameter name
- `value` = what you're assigning to it

```python
name="Max"   # key: name, value: "Max"
age=25       # key: age,  value: 25
```

This is literally the same shape as a dictionary entry — which is *why* `**kwargs` collects them into a dict.

```python
def f(**kwargs):
    print(kwargs)

f(name="Max", age=25)
# {'name': 'Max', 'age': 25}
```

## Side-by-side comparison

| Type | How it's matched | Example call |
|---|---|---|
| Positional | By order | `greet("Max", 25)` |
| Keyword | By name (key=value) | `greet(name="Max", age=25)` |

## Why this distinction matters

- Positional args are shorter to write but order-sensitive — easy to make silent mistakes.
- Keyword args are more explicit and safer, especially when a function has many parameters — this is why config-heavy code (like model inference calls, API wrappers) almost always uses keyword arguments.

**One-line summary:** positional = matched by slot/order, keyword = matched by name, and a keyword argument is literally a key-value pair — same concept as a dict entry."""
````

````python
#25

def outer():
    x=10
    def inner():
        print(x)
    inner()
outer()

"""inner() is a nested function defined inside outer(). Python allows inner functions to access variables from their enclosing (outer) function's scope — this is called a closure.
Step by step
outer() runs, creating x = 10 in its local scope.

inner() is defined inside outer(), so it has access to outer()'s variables.

When inner() is called, it looks for x:

Not found in inner()'s own local scope.

Found in the enclosing scope (outer()'s scope) — this is called the enclosing/nonlocal scope.

It prints 10."""
````

````python
"""# Python Exception Types — Notes

## Q27. What exception does each line raise?

```python
int("abc")
[1, 2, 3][10]
{"a": 1}["b"]
1 / 0
None.upper()
open("no_such_file.txt")
```

## Answers

| Code | Exception | Reason |
|---|---|---|
| `int("abc")` | `ValueError` | `"abc"` is a valid string type, but its value can't be converted to an integer. |
| `[1, 2, 3][10]` | `IndexError` | The list has only indexes `0, 1, 2`. Index `10` doesn't exist. |
| `{"a": 1}["b"]` | `KeyError` | The dictionary has only key `"a"`. Key `"b"` was never added. |
| `1 / 0` | `ZeroDivisionError` | Division by zero is mathematically undefined. |
| `None.upper()` | `AttributeError` | `None` has no attributes/methods at all — `.upper()` doesn't exist on it. |
| `open("no_such_file.txt")` | `FileNotFoundError` | The file path doesn't exist on disk. |

## Key concept: `ValueError` vs `TypeError`

- **ValueError** → correct type, but invalid value (`int("abc")` — string is fine, content isn't a number).
- **TypeError** → wrong type entirely (`"5" + 5` — can't add string and int).

## Key concept: `IndexError` vs `KeyError`

- **IndexError** → happens with sequences (list, tuple, string) when the position doesn't exist.
- **KeyError** → happens with dictionaries when the key doesn't exist.

## Key concept: literal indexing (no variable needed)

```python
[1, 2, 3][10]     # same as: a = [1,2,3]; a[10]
{"a": 1}["b"]      # same as: d = {"a":1}; d["b"]
```

Python allows indexing directly on a list/dict literal — no syntax error, just a runtime lookup failure if the index/key doesn't exist. In real code, this is almost always written using a variable first (`a[10]`, `d["b"]`), but behavior is identical.

## Key concept: `AttributeError` from `None`

`None.upper()` fails because `None` is not a string — it has zero methods. This error is extremely common in real code when a function silently returns `None` (e.g., missing `return` statement), and you then try to call a method on the result without checking.

## Quick memory table

| Exception | Trigger |
|---|---|
| `ValueError` | Right type, bad value |
| `TypeError` | Wrong type used in an operation |
| `IndexError` | Sequence index out of range |
| `KeyError` | Dict key doesn't exist |
| `AttributeError` | Object has no such method/attribute |
| `ZeroDivisionError` | Division by zero |
| `FileNotFoundError` | File path doesn't exist |"""
````
